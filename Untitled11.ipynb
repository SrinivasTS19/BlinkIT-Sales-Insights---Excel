is used to create a new column called event_sequence that assigns a unique sequential number to every row in your dataset.

2. DATA INTEGRITY AND DUPLICATE REMOVA

```python
print("\nDuplicate rows:", df.duplicated().sum())

```

# Remove exact duplicate records.


```python

df = df.drop_duplicates().copy()
```

```python
print("Shape after duplicate removal:", df.shape)

```

3. CLEAN COLUMN NAMES


```python
df.columns = df.columns.str.strip().str.lower()
```

```python
text_columns = [
    "employee_id",
    "interface",
    "task",
    "task_name",
    "start_page",
    "action",
    "target",
    "detail",
    "result",
    "source"
]
```

```python
for col in text_columns:
    df[col] = df[col].astype("string").str.strip()

```

```python
df["interface"] = df["interface"].str.upper()

```

```python
df["action"] = df["action"].str.strip().str.title()

```

```python
df["task"] = df["task"].str.upper()
```

```python
df["source"] = df["source"].str.title()

```

4. DATA TYPE CONVERSION


```python
df["age"] = pd.to_numeric(
    df["age"],
    errors="coerce"
).astype("Int64")
```

# Converted employee age to numeric.

```python
df["elapsed_s"] = pd.to_numeric(
    df["elapsed_s"],
    errors="coerce"
)
```

 Converted elapsed seconds to float.

```python
df["elapsed_s_was_missing"] = df["elapsed_s"].isna()

```

 Keep track of missing elapsed values.


```python
df["elapsed_s_filled"] = df["elapsed_s"].fillna(0.0)

```

 Preserve original missing values.
Create another column for calculations if needed.

# 5. DATE AND TIMESTAMP PROCESSING

```python
# STEP 1: Convert date into datetime format
# Original format: YYYY-MM-DD

df["date"] = pd.to_datetime(
    df["date"],
    format="%Y-%m-%d",
    errors="coerce"
)
```

 Dataset date format: DD-MM-YYYY


```python
# STEP 2: Convert timestamp into timedelta
# Original format: HH:MM:SS.mmm

df["timestamp_delta"] = pd.to_timedelta(
    df["timestamp"],
    errors="coerce"
)
```

The timestamp is MM:SS.s, not HH:MM:SS.
 Convert it into a timedelta.


```python
# STEP 3: Combine date and timestamp
# Create a unified datetime column.

df["datetime"] = (
    df["date"] + df["timestamp_delta"]
)
```

Create a unified datetime.
 This represents the recording date plus the
 session-relative timestamp, NOT actual wall-clock time.

```python
# STEP 4: Extract event date

df["event_date"] = df["datetime"].dt.normalize()

```

```python
# STEP 5: Extract day name

df["event_day"] = df["datetime"].dt.day_name()


# STEP 6: Extract month name

df["event_month"] = df["datetime"].dt.month_name()
```

```python
 #6. VERIFY DATE AND TIMESTAMP CONVERSION
# ============================================================

print("\nDate and Timestamp Preview:")

print(
    df[
        [
            "date",
            "timestamp",
            "timestamp_delta",
            "datetime",
            "event_date",
            "event_day",
            "event_month"
        ]
    ].head(10)
)


# Check missing values

print("\nMissing Date Values:")

print(
    df[
        [
            "date",
            "timestamp_delta",
            "datetime",
            "event_date",
            "event_day",
            "event_month"
        ]
    ].isnull().sum()
)
```

Create useful date/time attributes.

#7. SORT EVENTS


```python
df = df.sort_values(
    by=[
        "employee_id",
        "interface",
        "task",
        "event_sequence"
    ]
).reset_index(drop=True)

```

Use employee, interface, task and original event order.


# 8. CREATE EVENT FLAGS

```python
# Identify task lifecycle events.
df["is_task_start"] = (
    df["action"] == "Task_Start"
)

df["is_task_done"] = (
    df["action"] == "Task_Done"
)

df["is_task_giveup"] = (
    df["action"] == "Task_Giveup"
)

# Identify user interaction events.
# Lifecycle markers are excluded from interaction counts.
lifecycle_events = [
    "Interface_Start",
    "Interface_End",
    "Task_Start",
    "Task_Done",
    "Task_Giveup",
    "Session_End"
]

df["is_interaction"] = (
    ~df["action"].isin(lifecycle_events)
)

# Count clicks.
df["is_click"] = (
    df["action"] == "Click"
)

# Count navigation events.
df["is_navigation"] = (
    df["action"] == "App_Navigation"
)

# Count possible error events.
# No explicit Error action was found in the supplied dataset.
df["is_error_event"] = (
    df["action"].str.contains(
        "error",
        case=False,
        na=False
    )
)
```

# 9. TASK-LEVEL PERFORMANCE SUMMARY

```python
summary_records = []

group_columns = [
    "employee_id",
    "interface",
    "task"
]

for (employee, interface, task), group in df.groupby(
    group_columns,
    dropna=False,
    sort=False
):

    group = group.sort_values("event_sequence")

    # Find task start and end events.
    start_events = group[
        group["action"] == "Task_Start"
    ]

    done_events = group[
        group["action"] == "Task_Done"
    ]

    giveup_events = group[
        group["action"] == "Task_GiveUp"
    ]

    # Determine task outcome.
    if not done_events.empty:
        outcome = "Success"
        end_event = done_events.iloc[-1]

    elif not giveup_events.empty:
        outcome = "Failure"
        end_event = giveup_events.iloc[-1]

    else:
        outcome = "Incomplete"
        end_event = None

    # Calculate task duration.
    duration = np.nan

    if not start_events.empty and end_event is not None:

        start_event = start_events.iloc[0]

        start_elapsed = start_event["elapsed_s"]
        end_elapsed = end_event["elapsed_s"]

        if pd.notna(start_elapsed) and pd.notna(end_elapsed):

            duration = max(
                0.0,
                float(end_elapsed - start_elapsed)
            )

        # Fallback to timestamp difference if elapsed_s
        # is missing for either boundary event.
        elif (
            pd.notna(start_event["timestamp_delta"])
            and pd.notna(end_event["timestamp_delta"])
        ):

            duration = max(
                0.0,
                (
                    end_event["timestamp_delta"]
                    - start_event["timestamp_delta"]
                ).total_seconds()
            )

    # Count task interactions.
    total_actions = int(
        group["is_interaction"].sum()
    )

    total_clicks = int(
        group["is_click"].sum()
    )

    total_navigation = int(
        group["is_navigation"].sum()
    )

    total_errors = int(
        group["is_error_event"].sum()
    )

    # Task completion indicator.
    success_flag = (
        1 if outcome == "Success" else 0
    )

    # Get task name and participant age.
    task_name = group["task_name"].iloc[0]
    age = group["age"].iloc[0]

    summary_records.append({

        "employee_id": employee,
        "age": age,
        "interface": interface,
        "task": task,
        "task_name": task_name,

        "task_outcome": outcome,
        "success_flag": success_flag,

        "task_duration_s": duration,

        "total_actions": total_actions,
        "total_clicks": total_clicks,
        "total_navigation": total_navigation,
        "error_event_count": total_errors,

        "task_start_count": len(start_events),
        "task_done_count": len(done_events),
        "task_giveup_count": len(giveup_events),

        "total_event_count": len(group)

    })


# Convert summary records to DataFrame.
summary = pd.DataFrame(summary_records)

```

# 10. ADD SESSION-LEVEL METRICS


```python
# In this dataset, employee + interface represents a session.
# The task-level summary is grouped by employee/interface/task.

session_metrics = (
    summary.groupby(
        ["employee_id", "interface"]
    )
    .agg(
        session_action_count=("total_actions", "sum"),
        session_click_count=("total_clicks", "sum"),
        session_navigation_count=(
            "total_navigation", "sum"
        ),
        session_task_count=("task", "nunique"),
        session_success_count=("success_flag", "sum")
    )
    .reset_index()
)

# Add session metrics to each task summary row.
summary = summary.merge(
    session_metrics,
    on=["employee_id", "interface"],
    how="left"
)

```

# 11. ADD COMPARISON-FRIENDLY METRICS


```python
summary["task_success_rate"] = (
    summary["success_flag"] * 100
)

summary["task_failed_flag"] = (
    summary["task_outcome"] == "Failure"
).astype(int)

# A completed task has a valid measured duration.
summary["duration_available"] = (
    summary["task_duration_s"].notna()
).astype(int)

```

# 12. FINAL DATA QUALITY CHECKS

```python
print("\n========== DATA QUALITY REPORT ==========")

print("Cleaned event rows:", len(df))

print("Duplicate rows remaining:", df.duplicated().sum())

print("\nMissing values:")
print(df.isna().sum())

print("\nInterface distribution:")
print(df["interface"].value_counts())

print("\nAction distribution:")
print(df["action"].value_counts())

print("\nTask outcome distribution:")
print(summary["task_outcome"].value_counts())

```

# 13. EXPORT CLEANED DATASETS

```python
# Export the complete event-level dataset.
df.to_csv(
    "cleaned_interaction_logs.csv",
    index=False
)

# Export the task/session performance summary.
summary.to_csv(
    "session_summary.csv",
    index=False
)

print("\n========================================")
print("DATA CLEANING COMPLETED SUCCESSFULLY")
print("========================================")

print("Created: cleaned_interaction_logs.csv")
print("Created: session_summary.csv")

print("\nSummary dataset shape:", summary.shape)
print("\nSummary preview:")
print(summary.head())
```
